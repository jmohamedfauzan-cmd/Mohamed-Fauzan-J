STACK

Operations:
 
1.Push()
2.Pop()
3.Peek()
4.isEmpty()

```python
stack = []
stack.append(10)  # push
stack.append(20)
stack.append(30)

print(stack)  # [10, 20, 30]

stack.pop()  # pop
print(stack[-1])  # peek

print(len(stack))
```

```python
class MinStack:
    def __init__(self):
        self.stack = []
        self.min_stack = []

    def push(self, value):
        self.stack.append(value)
        if not self.min_stack or value <= self.min_stack[-1]:
            self.min_stack.append(value)
    
    def pop(self):
        if not self.stack:
            return "Stack is empty"
        value = self.stack.pop()
        if value == self.min_stack[-1]:
            self.min_stack.pop()
        return value

    def get_min(self):
        if not self.min_stack:
            return "Stack is empty"
        return self.min_stack[-1]

s = MinStack()
s.push(10)
s.push(5)
s.push(20)
s.push(2)
print("Minimum:",s.get_min())
s.pop()
print("Minimum after pop:",s.get_min())
```
